### Consolidación — Streamlit

Con Streamlit ya sabes construir una aplicación web completa usando solo
Python. Los cinco pilares que has practicado son:

1.  Modelo de ejecución: el script se re-ejecuta de arriba abajo en cada
    interacción.
2.  Componentes interactivos: `slider`, `selectbox`, `multiselect`,
    `checkbox`.
3.  Layout: `sidebar` para controles, `tabs` para organización,
    `columns` para espacio horizontal.
4.  Cacheo: `@st.cache_data` para operaciones costosas.
5.  Integración con Plotly: `st.plotly_chart(fig)` en lugar de
    `fig.show()`.

El siguiente esqueleto recoge ese patrón completo — úsalo como plantilla
de partida en cualquier proyecto real:

Streamlit es la elección natural para dashboards internos de equipo,
prototipos rápidos para validar ideas con *stakeholders*, demos de
modelos y herramientas de exploración para usuarios no técnicos.

**Cuándo NO usar Streamlit:**

-   Aplicaciones web empresariales con millones de usuarios
-   Apps que requieren autenticación compleja y roles de usuario
-   Servicios web con APIs RESTful
-   Aplicaciones que necesitan control fino de frontend (CSS/JS
    complejo)
